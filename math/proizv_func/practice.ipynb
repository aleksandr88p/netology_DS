# производные 
    это мгновенная скорость изменения функции
    в тоже время это тоже функция

    ряд простых правил вычисления производных функций
    1) если есть постоянное значение например игрек равен 5 то производная равна 0
    d/dx(5) = 0
    2) если функция равная некоторой ПЕРЕМЕННОЙ велечине например игрек равен икс, то производная равна 1
    d/dx(x) = 1
    3) если встречатется произведение двух первых функций, например игрек равен 5 умножить на икс, то производная равна 5
    d/dx(5*x) = 5
    4) с нелинейными функциями(тоесть икс в какой то степени) y = 3x^4, то производная равна 12x^3
    d/dx(3x^4) = 12x^3

# отдельные правила для производных тригонометрических функций
    d/dx(sin(x)) = cos(x)
    d/dx(cos(x)) = -sin(x)
    d/dx(tan(x)) = sec^2(x)
    d/dx(cot(x)) = -csc^2(x)
    d/dx(sec(x)) = sec(x)tan(x)
    d/dx(csc(x)) = -csc(x)cot(x)




# правило для производных сложных функций
    d/dx(f(g(x))) = f'(g(x)) * g'(x)
    d/dx(sin(x^2)) = cos(x^2) * 2x
    d/dx(e^(3x)) = e^(3x) * 3
    d/dx(ln(x^2)) = 2/x
    d/dx(sqrt(x^3)) = 3/2 * x^(1/2)
    d/dx(sin(x^2 + 3x)) = cos(x^2 + 3x) * (2x + 3)

```python
# Нужные библиотеки
import numpy as np
import matplotlib.pyplot as plt
import sympy as sym

# Создаём функцию
def f(x):
    return x**2 + 2*x + 1

# Создаём точки для графика
x = np.linspace(-5, 5, 100)  # 100 точек от -5 до 5
y = f(x)

# Строим график
plt.plot(x, y, label='f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Символьная переменная
x = sym.Symbol('x')
# Задаём функцию
f = x**2 + 2*x + 1
# Находим производную
f_prime = sym.diff(f, x)

print(f"Производная: {f_prime}")
# Выведет: 2*x + 2
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Несколько функций
def f1(x): return x**2        # квадратичная
def f2(x): return 2*x + 1     # линейная
def f3(x): return np.sin(x)   # синус

# Точки для графика
x = np.linspace(-5, 5, 100)

# Строим все графики
plt.plot(x, f1(x), label='x²')
plt.plot(x, f2(x), label='2x+1')
plt.plot(x, f3(x), label='sin(x)')

plt.grid(True)
plt.legend()
plt.title('Разные функции')
plt.show()
```
