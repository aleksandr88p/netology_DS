```python
# работа со случайными величинами на языке python
import numpy as np
```

```python
# экспоненциальное распределение
np.random.exponential(scale=2.0, size=(3,))
```

```python
# нормальное распределение
# loc среднее значение, в данном случае 0
# scale стандартное отклонение, в данном случае 1(разброс)
tmp = np.random.normal(loc=0.0, scale=1.0, size=(8000,))
tmp
```

```python
tmp.mean() # среднее значение
```

```python
tmp.std() # среднеквадратичное отклонение
```

```python
np.median(tmp) # медиана
```
