## Задание 4


Дана функция с неизвестными параметрами a и b, вам также известны значения функции `fx` в точках `x`. 

Найдите оптимальные параметры a, b, минимизирующие абсолютную ошибку `error`.


```python
def f(x, a, b):
    return np.e**a * np.sin(b*x) + x
```

```python
x = np.array([0.        , 0.26315789, 0.52631579, 0.78947368, 1.05263158,
              1.31578947, 1.57894737, 1.84210526, 2.10526316, 2.36842105,
              2.63157895, 2.89473684, 3.15789474, 3.42105263, 3.68421053,
              3.94736842, 4.21052632, 4.47368421, 4.73684211, 5.        ])

fx = np.array([  0.        , -12.01819092, -18.90968634, -17.68786571,
                -8.7529108 ,   4.27524517,  16.06801336,  21.81250213,
                19.22059845,   9.48411207,  -3.22273056, -13.48576488,
               -16.91096359, -11.95866834,  -0.58630088,  12.56873816,
                22.12489421,  24.20292139,  18.04522521,   6.33211092])
```

```python
plt.plot(x, fx)
plt.show()
```

```python
# def error(params):
#     return np.sum(np.abs(fx - f(x, params[0], params[1])))

def error(params):
    current_error = np.sum(np.abs(fx - f(x, params[0], params[1])))
    print(f"Текущие параметры: a = {params[0]}, b = {params[1]}, Ошибка = {current_error}")
    return current_error
```

```python
### YOUR CODE HERE
initial_params = [1, 1]

# Оптимизация
result = minimize(error, initial_params)

# Оптимальные параметры
optimal_a, optimal_b = result.x
print(f"Оптимальные параметры: a = {optimal_a}, b = {optimal_b}")

# Визуализация
plt.plot(x, fx, label='первые значения fx')
plt.plot(x, f(x, optimal_a, optimal_b), label='Оптимизированная функция', linestyle='--')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize, curve_fit, differential_evolution



# Вариант 1: curve_fit
params, _ = curve_fit(f, x, fx, p0=[1, 1])
optimal_a_1, optimal_b_1 = params

# Вариант 2: COBYLA
def error(params):
    return np.sum(np.abs(fx - f(x, params[0], params[1])))

result = minimize(error, [1, 1], method='COBYLA')
optimal_a_2, optimal_b_2 = result.x

# Вариант 3: differential_evolution
bounds = [(-10, 10), (-10, 10)]  
result_de = differential_evolution(error, bounds)
optimal_a_3, optimal_b_3 = result_de.x

print("Результаты curve_fit:")
print(f"a = {optimal_a_1}, b = {optimal_b_1}")

print("\nРезультаты кобыла:")
print(f"a = {optimal_a_2}, b = {optimal_b_2}")

print("\nРезультаты differential evolution:")
print(f"a = {optimal_a_3}, b = {optimal_b_3}")

# Визуализация
plt.figure(figsize=(12, 6))
plt.plot(x, fx, label='Исходные данные', marker='o')
plt.plot(x, f(x, optimal_a_1, optimal_b_1), label='curve_fit', linestyle='--')
plt.plot(x, f(x, optimal_a_2, optimal_b_2), label='кобыла', linestyle=':')
plt.plot(x, f(x, optimal_a_3, optimal_b_3), label='differential_evolution', linestyle='-.')
plt.legend()
plt.grid(True)
plt.show()
```

Дана матрица рейтингов фильмов (строки - пользователи (10 человек), столбцы - фильмы (15 фильмов))

```python
raitings = np.array([[ 4,  4,  9,  4,  1,  6, 10,  7,  9,  6,  9,  2,  8,  6,  6],
                     [ 9,  2,  5, 10,  7,  8, 10,  5,  6,  2,  1,  6,  8,  9,  7],
                     [ 1,  6,  8,  8,  4,  9,  3,  8, 10,  5,  2,  6,  8,  1,  6],
                     [ 6,  1,  9,  7,  7,  9,  2,  3,  5,  1,  6,  6,  3,  2,  7],
                     [ 3,  7,  3,  5,  7,  9,  9,  6,  2,  9,  1,  2,  8, 10,  6],
                     [ 8,  3,  7,  3,  8,  6,  1,  8,  8,  6,  1,  9,  4, 10,  1],
                     [ 9,  8,  4,  8,  8, 10,  6,  1,  1,  2,  9,  5,  2,  7,  2],
                     [ 4,  1,  6,  4,  3, 10,  4,  4,  2,  8,  7,  9,  3,  8,  3],
                     [ 2,  7,  7,  6, 10,  6,  8,  9,  8,  6, 10,  1,  7, 10,  4],
                     [ 5, 10,  8,  8,  9,  7,  2,  9,  9, 10,  8,  8,  8,  6, 10]])
```

Необходимо найти наиболее похожего пользователя по косинусной метрике для каждого из пользователей (ответ: 10 пар вида (x, y), где y - наиболее похожий пользователь на пользователя x).

```python
### YOUR CODE HERE
import numpy as np
from sklearn.metrics import pairwise_distances

# Вычисление косинусного расстояния
cosine_distances = pairwise_distances(raitings, metric='cosine')

# Нахождение наиболее похожих пользователей
most_similar_users = []
for i in range(cosine_distances.shape[0]):
    # индексы пользователей, отсортированных по расстоянию
    similar_indices = np.argsort(cosine_distances[i])
    # пользователь и наиболее похожий пользователь
    most_similar_users.append((i, similar_indices[1]))

for user, similar_user in most_similar_users:
    print(f"Пользователь {user} наиболее похож на пользователя {similar_user}")
```
