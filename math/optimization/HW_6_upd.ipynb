```python

def f(x, a, b):
    return np.e**a * np.sin(b*x) + x

### YOUR CODE HERE
best_error = float('inf')  # начальная ошибка 
best_params = None  # лучшие параметры
best_p0 = None  # лучшие начальные 

for a_start in [-1, 10, 1]:
    for b_start in [0.5, 10, 2]:
        try:
            res = curve_fit(f, x, y, p0=[a_start, b_start])
            error = np.mean((y - f(x, res[0][0], res[0][1]))**2)
            
            # Сохраняю
            if error < best_error:
                best_error = error
                best_params = res[0]
                best_p0 = [a_start, b_start]
            
            print(f"Начальные параметры: a={a_start}, b={b_start}")
            print(f"Найденные параметры: a={res[0][0]:.4f}, b={res[0][1]:.4f}")
            print(f"Ошибка: {error:.4f}\n")
        except:
            print(f"Ошибка при a={a_start}, b={b_start}\n")

# лучшие найденные параметры
print("the best параметры:")
print(f"Начальные параметры: a={best_p0[0]}, b={best_p0[1]}")
print(f"Найденные параметры: a={best_params[0]:.4f}, b={best_params[1]:.4f}")
print(f"Минимальная ошибка: {best_error:.4f}")

# Сохраняем лучший результат 
res = curve_fit(f, x, y, p0=best_p0)
```
